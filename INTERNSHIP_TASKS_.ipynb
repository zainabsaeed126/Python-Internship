1 Reverse a string without using a built-in reverse function.

```python
text=input("Enter a string: ")

reverse_text=" "
for char in text:
  reverse_text=char+reverse_text

  print(reverse_text)
```

4 Create a Student class with name/marks attributes and a method to compute grade.

```python
class student:

  def __init__(self, name, marks):
    self.name=name
    self.marks=marks
  def grade(self):
    if self.marks >=80:
      return "A+"
      if self.marks >=70:
        return "A"
      elif self.marks >=60:
          return "B"
      elif self.marks >=50:
            return "C"

name=input("Enter student name: ")
marks=int(input("Enter student marks: "))

student= student (name, marks)

print("Name:", student.name)
print("Marks:", student.marks)
print("grade:", student.grade())
```

5 Handle division-by-zero and invalid input safely with try/except.

```python
try:
  num1 = int(input("Enter a number: "))
  num2 = int(input("Enter another number: "))

  result = num1/num2

  print("Result:", result)

except ZeroDivisionError:
  print("Error! cannot divide by zero")

except ValueError:
  print("Error! enter a valid number")

```

6 Simple To-Do List app (add/remove/view tasks) using functions and a list.

```python
tasks = []

def add_task():
    task = input("Enter task: ")
    tasks.append(task)
    print("Task added.")

def remove_task():
    task = input("Enter task to remove: ")

    if task in tasks:
        tasks.remove(task)
        print("Task removed.")
    else:
        print("Task not found.")

def view_tasks():
    if len(tasks) == 0:
        print("No tasks.")

    else:
        print("Your Tasks:")
        for task in tasks:
            print(task)


while True:

    print("\n1. Add Task")
    print("2. Remove Task")
    print("3. View Tasks")
    print("4. Exit")

    choice = input("Enter your choice: ")

    if choice == "1":
        add_task()

    elif choice == "2":
        remove_task()

    elif choice == "3":
        view_tasks()

    elif choice == "4":
        print("Goodbye!")
        break

    else:
        print("Invalid choice.")
```
