1 Reverse a string without using a built-in reverse function.

```python
text=input("Enter a string: ")

reverse_text=" "
for char in text:
  reverse_text=char+reverse_text

  print(reverse_text)
```

2 Check if two strings are anagrams of each other.

```python
a=input("Enter a string: ")
b=input("Enter a string: ")

if sorted(a)==sorted(b):
  print("Anagrams")
else:
  print("Not Anagrams")
```

4 Create a Student class with name/marks attributes and a method to compute grade.

```python
class student:

  def __init__(self, name, marks):
    self.name=name
    self.marks=marks
  def grade(self):
    if self.marks >=80:
      return "A+"
      if self.marks >=70:
        return "A"
      elif self.marks >=60:
          return "B"
      elif self.marks >=50:
            return "C"

name=input("Enter student name: ")
marks=int(input("Enter student marks: "))

student= student (name, marks)

print("Name:", student.name)
print("Marks:", student.marks)
print("grade:", student.grade())
```

5 Handle division-by-zero and invalid input safely with try/except.

```python
try:
  num1 = int(input("Enter a number: "))
  num2 = int(input("Enter another number: "))

  result = num1/num2

  print("Result:", result)

except ZeroDivisionError:
  print("Error! cannot divide by zero")

except ValueError:
  print("Error! enter a valid number")

```
