**Week 2 - Mini Project
Data Cleaning Tool**

```python
import pandas as pd

# Read CSV
df = pd.read_csv("messy_data.csv")

# Remove duplicate rows
df = df.drop_duplicates()

# Fill missing AGE with 0
df["AGE"] = df["AGE"].fillna(0)

# Fill missing CITY with Unknown
df["CITY"] = df["CITY"].fillna("Unknown")

# Save cleaned file
df.to_csv("cleaned_data.csv", index=False)

# Print summary and display cleaned data
print("Cleaning Completed: Duplicates Removed, Missing AGE Filled, Missing CITY Filled")
display(df.head())
```
