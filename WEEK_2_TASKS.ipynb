**Task 1: Read & Write CSV and JSON + Filter Records**

```python
import csv

# Read CSV file
with open("students.csv", "r") as file:
    data = csv.reader(file)

    for row in data:
        print(row)

# Write CSV file
with open("new_students.csv", "w", newline="") as file:
    writer = csv.writer(file)

    writer.writerow(["Name", "Age"])
    writer.writerow(["Ali", 20])
    writer.writerow(["Sara", 21])
```

```python
import json

data = [
    {"name": "Ali", "age": 20},
    {"name": "Sara", "age": 21}
]

# Write JSON
with open("students.json", "w") as file:
    json.dump(data, file)

# Read JSON
with open("students.json", "r") as file:
    students = json.load(file)

for student in students:
    if student["age"] > 20:
        print(student)
```

**Task 2: NumPy Arrays, Slicing & Basic Statistics**

```python
import numpy as np

arr = np.array([10, 20, 30, 40, 50])

print(arr)

print(arr[1:4])

print("Sum =", np.sum(arr))
print("Mean =", np.mean(arr))
print("Max =", np.max(arr))
print("Min =", np.min(arr))
```

**Task 3: Pandas (Load, Filter, Group, Missing Values)**

```python
import pandas as pd

data = {
    "Name": ["Ali", "Sara", "Ali", "Ahmed"],
    "Marks": [80, 90, None, 70]
}

df = pd.DataFrame(data)

print(df)

# Fill missing values
df["Marks"] = df["Marks"].fillna(0)

# Filter
print(df[df["Marks"] > 75])

# Group By
print(df.groupby("Name")["Marks"].mean())
```
