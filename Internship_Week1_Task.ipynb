Task 1: Swap two numbers without a third variable

```python
a = 56
b = 23

a, b = b, a
# this will swap the values

print("a =", a)
print("b =", b)
```

Palindrome and Prime Number Chekcing


```python
#__________________Prime_Number__________________________________
num = int(input("Enter a number: "))

if num > 1:
    for i in range(2, num):
        if (num % i) == 0:
            print(num, "is not a prime number.")
            break
    else:
        print(num, "is a prime number.")

else:
    print(num, "is not a prime number.")

#________________________palindrome__________________________________
a = input("Enter a string: ")

if a == a[::-1]:
   print("palindrome")
else:
  print("not a palindrome")
```
