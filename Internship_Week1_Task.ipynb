Task 2: Palindrome and Prime Number Chekcing


```python
#__________________Prime_Number__________________________________
num = int(input("Enter a number: "))

if num > 1:
    for i in range(2, num):
        if (num % i) == 0:
            print(num, "is not a prime number.")
            break
    else:
        print(num, "is a prime number.")

else:
    print(num, "is not a prime number.")

#________________________palindrome__________________________________
a = input("Enter a string: ")

if a == a[::-1]:
   print("palindrome")
else:
  print("not a palindrome")
```

Task 3: Fabonacci serirs by loop and recurssion

```python
a = int(input("Enter number of terms: "))
print("\nfibonacci series using loop")

x=0
y=1
# x is first number
# y is second number

for i in range(a):
  print(x, end=" ")
  z = x+y
  x = y
  y = z

  #____________________using recursion_____________________
n = int(input("\nEnter number of terms: "))
print("\n\nFibonacci Series using Recursion:")

def fibonacci(n):
    if n <= 1:
        return n
    else:
        return fibonacci(n - 1) + fibonacci(n - 2)

for i in range(n):
    print(fibonacci(i), end=" ")


```

Task 4: function library:

```python
def celcius_to_farenheit(c):
  return(c * 9/5) + 32

def farenheit_to_celcius(f):
  return(f - 32) * 5/9

def circle_area(r):
   return 3.14 * r * r

def rectangle_area(l, w):
   return l * w

print(celcius_to_farenheit(23))
print(farenheit_to_celcius(56))
print(circle_area(8))
print(rectangle_area(8, 5))
```
