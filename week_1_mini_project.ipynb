```python
def calculate_grade(percentage):

    if percentage >= 90:
        return "A+"

    elif percentage >= 80:
        return "A"

    elif percentage >= 70:
        return "B"

    elif percentage >= 60:
        return "C"

    elif percentage >= 50:
        return "D"

    else:
        return "F"


try:

    obtained = float(input("Enter obtained marks: "))
    total = float(input("Enter total marks: "))

    if obtained < 0 or total <= 0:
        print("Marks cannot be negative or zero.")

    elif obtained > total:
        print("Obtained marks cannot be greater than total marks.")

    else:
        percentage = (obtained / total) * 100

        grade = calculate_grade(percentage)

        print("Percentage =", round(percentage, 2), "%")
        print("Grade =", grade)

except ValueError:
    print("Invalid input! Please enter numbers only.")
```
