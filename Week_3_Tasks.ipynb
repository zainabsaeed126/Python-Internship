**Plot Bar Chart, Line Chart and Pie Chart**

```python
import matplotlib.pyplot as plt

subjects = ["Math", "English", "Science", "Computer"]
marks = [80, 70, 90, 85]

# Bar Chart
plt.bar(subjects, marks)
plt.title("Bar Chart")
plt.show()

# Line Chart
plt.plot(subjects, marks)
plt.title("Line Chart")
plt.show()

# Pie Chart
plt.pie(marks, labels=subjects, autopct="%1.1f%%")
plt.title("Pie Chart")
plt.show()
```

**Week 3 - Task 2
Correlation Heatmap Using Seaborn**

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

data = {
    "Math": [80, 70, 90, 85],
    "English": [75, 65, 88, 80],
    "Science": [90, 80, 95, 85]
}

df = pd.DataFrame(data)

sns.heatmap(df.corr(), annot=True)

plt.title("Correlation Heatmap")
plt.show()
```

**Week 3 - Task 3
Fetch Data from a Public REST API and Save to File**

```python
import requests
import json

url = "https://jsonplaceholder.typicode.com/users"

response = requests.get(url)

data = response.json()

with open("users.json", "w") as file:
    json.dump(data, file, indent=4)

print("Data saved successfully.")
```

** Week 3 — Task 4
Parse a JSON API response and turn part of it into a chart**

```python
import requests
import matplotlib.pyplot as plt

# API se data lena
url = "https://jsonplaceholder.typicode.com/posts"

response = requests.get(url)

# JSON ko Python data mein convert karna
data = response.json()

# Har user ke posts count karna
user_ids = []

for post in data:
    user_ids.append(post["userId"])

# First 5 users ka data lena
users = []
counts = []

for user in range(1, 6):
    users.append(user)
    counts.append(user_ids.count(user))

# Bar chart banana
plt.bar(users, counts)

plt.title("Posts by Users")
plt.xlabel("User ID")
plt.ylabel("Number of Posts")

plt.show()
```

**Week 3 — Task 5
Create 2–3 reusable plotting functions**

```python
import matplotlib.pyplot as plt

# Bar chart function
def bar_chart(x, y):
    plt.bar(x, y)
    plt.title("Bar Chart")
    plt.show()


# Line chart function
def line_chart(x, y):
    plt.plot(x, y)
    plt.title("Line Chart")
    plt.show()


# Pie chart function
def pie_chart(values, labels):
    plt.pie(values, labels=labels, autopct="%1.1f%%")
    plt.title("Pie Chart")
    plt.show()


# Example data
subjects = ["Math", "English", "Science", "Computer"]
marks = [80, 70, 90, 85]

# Use functions
bar_chart(subjects, marks)

line_chart(subjects, marks)

pie_chart(marks, subjects)
```
