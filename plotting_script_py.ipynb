```python
import matplotlib.pyplot as plt

# Data
subjects = ["Math", "English", "Science", "Computer"]
marks = [80, 70, 90, 85]

# Bar chart
plt.bar(subjects, marks)
plt.title("Student Marks")
plt.xlabel("Subjects")
plt.ylabel("Marks")
plt.show()
```

```python
code = """
import matplotlib.pyplot as plt

subjects = ["Math", "English", "Science"]
marks = [80, 70, 90]

plt.bar(subjects, marks)
plt.title("Student Marks")
plt.show()
"""

with open("plotting_script.py", "w") as file:
    file.write(code)

print("File updated")
```
